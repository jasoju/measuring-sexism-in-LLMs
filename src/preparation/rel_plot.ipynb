```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# ASI

af=pd.read_json(f"../../results/ASI/rel_eval_af.json", orient="index")
r=pd.read_json(f"../../results/ASI/rel_eval_reversed.json", orient="index")
eos=pd.read_json(f"../../results/ASI/rel_eval_changed_eos.json", orient="index")

df_ASI = pd.concat([af,r["reversed"],eos["changed_eos"]], axis=1)
df_ASI["model"] = ["Llama 3.1 70B", 
                   "Llama 3.1 8B", 
                   "Cenatur", 
                   "Llama 3.3 70B", 
                   "Mistral 7B v0.3", 
                   "Qwen 2.5 14B",
                   "Qwen 2.5 32B",
                   "Qwen 2.5 7B",
                   "Qwen 3 4B",
                   "Gemma 3 12B",
                   "Gemma 3 1B",
                   "Gemma 3 27B",
                   "Gemma 3 4B"
                   ]
df_ASI["order"] = [7,
                   6,
                   1,
                   8,
                   9,
                   11,
                   12,
                   10,
                   13,
                   4,
                   2,
                   5,
                   3]
df_ASI["test"] = "ASI"

df_ASI = df_ASI.sort_values(by=["order"], ignore_index=True)
```

```python
# SR2K

af=pd.read_json(f"../../results/SR2K/rel_eval_af.json", orient="index")
r=pd.read_json(f"../../results/SR2K/rel_eval_reversed.json", orient="index")
eos=pd.read_json(f"../../results/SR2K/rel_eval_changed_eos.json", orient="index")

df_SR2K = pd.concat([af,r["reversed"],eos["changed_eos"]], axis=1)
df_SR2K["model"] = ["Llama 3.1 70B", 
                   "Llama 3.1 8B", 
                   "Cenatur", 
                   "Llama 3.3 70B", 
                   "Mistral 7B v0.3", 
                   "Qwen 2.5 14B",
                   "Qwen 2.5 32B",
                   "Qwen 2.5 7B",
                   "Qwen 3 4B",
                   "Gemma 3 12B",
                   "Gemma 3 1B",
                   "Gemma 3 27B",
                   "Gemma 3 4B"
                   ]
df_SR2K["order"] = [7,
                   6,
                   1,
                   8,
                   9,
                   11,
                   12,
                   10,
                   13,
                   4,
                   2,
                   5,
                   3]
df_SR2K["test"] = "SR2K"

df_SR2K = df_SR2K.sort_values(by=["order"], ignore_index=True)
```

```python
# MFQ

af=pd.read_json(f"../../results/MFQ/rel_eval_af.json", orient="index")
r=pd.read_json(f"../../results/MFQ/rel_eval_reversed.json", orient="index")
eos=pd.read_json(f"../../results/MFQ/rel_eval_changed_eos.json", orient="index")

df_MFQ = pd.concat([af,r["reversed"],eos["changed_eos"]], axis=1)
df_MFQ["model"] = ["Llama 3.1 70B", 
                   "Llama 3.1 8B", 
                   "Cenatur", 
                   "Llama 3.3 70B", 
                   "Mistral 7B v0.3", 
                   "Qwen 2.5 14B",
                   "Qwen 2.5 32B",
                   "Qwen 2.5 7B",
                   "Qwen 3 4B",
                   "Gemma 3 12B",
                   "Gemma 3 1B",
                   "Gemma 3 27B",
                   "Gemma 3 4B"
                   ]
df_MFQ["order"] = [7,
                   6,
                   1,
                   8,
                   9,
                   11,
                   12,
                   10,
                   13,
                   4,
                   2,
                   5,
                   3]
df_MFQ["test"] = "MFQ"

df_MFQ = df_MFQ.sort_values(by=["order"], ignore_index=True)
```

```python
df_all = pd.concat([df_ASI, df_SR2K, df_MFQ])
```

```python
print(sns.color_palette("colorblind").as_hex())
```

```python
sns.set_style("darkgrid")
sns.color_palette("colorblind6")

fig, axs = plt.subplots(ncols=3, sharey=True, sharex=True, figsize=(10, 4.5))
sns.scatterplot(x='af', y='model', data=df_all, ax=axs[0], hue="test", legend=True, zorder=2, palette=sns.color_palette("colorblind"))
sns.scatterplot(x='reversed', y='model', data=df_all, ax=axs[1], hue="test", legend=False, zorder=2, palette=sns.color_palette("colorblind"))
sns.scatterplot(x='changed_eos',y='model', data=df_all, ax=axs[2], hue="test", legend=False, zorder=2, palette=sns.color_palette("colorblind"))

#fig.supxlabel("Fraction of consistent answers", fontsize=12)
axs[0].set_ylabel("Model", fontsize=12)
#axs[0].set_xlim(0,1)

axs[0].set_title("Alternate form", fontsize=12)
axs[1].set_title("Reversed order", fontsize=12)
axs[2].set_title("Changed EOS", fontsize=12)

axs[0].axvline(0.54, c="#0173b2", ls="--", lw=1,zorder=1)
axs[0].axvline(0.83, c="#de8f05", ls="--", lw=1,zorder=1)
axs[0].axvline(0.6, c="#029e73", ls="--", lw=1,zorder=1)
axs[1].axvline(1, c="grey", ls="--", lw=1,zorder=1)
axs[2].axvline(1, c="grey", ls="--", lw=1,zorder=1)

axs[0].set(xlabel=None)
axs[1].set_xlabel("Fraction of consistent answers", fontsize=12)
axs[2].set(xlabel=None)

# Get handles and labels from the first plot
handles, labels = axs[0].get_legend_handles_labels()
axs[0].get_legend().remove()

# Add one shared legend to the whole figure
fig.legend(handles, labels, bbox_to_anchor=(0.84, 0.939), loc=2)

plt.tight_layout(rect=[0, 0, 0.85, 1])  # leave space for legend
```
